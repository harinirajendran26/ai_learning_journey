```python
def is_even(n):
  if n%2 == 0:
      return True
  else:
       return False

print(is_even(3))   # False
print(is_even(4))   # True
print(is_even(0))   # True
```

Object Oriented Programming: lets you bundle data with behaviour(functions) into a single unit called Object.

Class - blueprint
object - instant built from it


class Dog: defines the blueprint

__init__ is the special "constructor" method — it runs automatically when you create an object, setting up its initial data

self refers to "this specific object" — every method needs it as the first parameter, so it knows which object's data to use

self.name = name stores name as an attribute on this object, so it's remembered even after __init__ finishes

```python
class Student:
    def __init__(self,name,grade):
        self.name = name
        self.grade = grade

    def show_info(self):
        print(f"{self.name} is in grade {self.grade}")

    def update_grade(self,new_grade):
        self.grade = new_grade

Student1 = Student("Alice", "A")
Student1.show_info()
Student1.update_grade("A+")
Student1.show_info()  

student2 = Student("Ravi", "B")
student2.show_info()
Student1.show_info()   
```
